```
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import time

st.title("🛡️ IoT Gas Turbine Health Monitor")
st.write("Simulating Real-Time Sensor Data from NASA C-MAPSS")

# 1. Simulate the AI prediction logic
# (In your real app, you would load your 'model.pkl' here)
def predict_rul(sensor_val):
    # This is a dummy logic for demonstration
        # It mimics health dropping as sensor value increases
            return max(0, 200 - (sensor_val - 500) * 10)

            # 2. Setup Dashboard Layout
            col1, col2 = st.columns(2)
            health_gauge = col1.empty()
            status_indicator = col2.empty()
            chart_placeholder = st.empty()

            # 3. Data Loop (The "IoT" Simulation)
            history = []
            for i in range(100):
                # Simulating a sensor reading (e.g., Temperature)
                    fake_sensor_temp = 500 + i + np.random.normal(0, 2)
                        predicted_life = predict_rul(fake_sensor_temp)

                                # Update Gauge
                                    health_gauge.metric("Remaining Useful Life", f"{int(predicted_life)} Cycles")

                                            # Update Status
                                                if predicted_life > 100:
                                                        status_indicator.success("ENGINE STATUS: STABLE")
                                                            elif predicted_life > 30:
                                                                    status_indicator.warning("ENGINE STATUS: WARNING")
                                                                        else:
                                                                                status_indicator.error("ENGINE STATUS: DANGER")

                                                                                        # Update Chart
                                                                                            history.append(fake_sensor_temp)
                                                                                                chart_placeholder.line_chart(history)

                                                                                                        time.sleep(0.5)
```
